###다양한 Matplotlib 사용방법

Pyplot 방식

```python
import matplotlib.pyplot as plt
x = df.index
y1 = df.국어
y2 = df.영어

plt.plot(x,y1,'r-o',lw=2, label='국어')
plt.plot(x,y2,'b--x',lw=1, label='영어')
plt.legend()
```

객체지향 API 그림그리기


```python
import matplotlib.pyplot as plt

x = df.index
y1 = df.국어
y2 = df.영어

fig = plt.figure()   # 직접 Figure 객체를 생성
ax = fig.subplots()   # 직접 axes를 생성
ax.plot(x,y1,'r-*',lw=1)    # 생성된 axes에 대한 plot() 멤버 직접 호출
ax.plot(x,y2,'b--o',lw=1)
```

두 가지 방식을 조합하여 Figure와 Axes를 plt.subpolots()라는 편의 함수를 사용

```python
import matplotlib.pyplot as plt

x = df.index
y1 = df.국어
y2 = df.영어

fig,ax = plt.subplots()
# plt.subplots() 편의 함수는 Figure 객체를 생성하고
# Figure.subplots()를 호출하여 리턴
ax.plot(x,y1,'r-*',lw=1)
ax.plot(x,y2,'b--o',lw=1)
```

### 도면을 나누어서 그리기.

```python
# Method1, 파이플롯 스타일, subplot()
fig = plt.figure(figsize=(5,4)) #그림 크기 지정이 필요없으면 생략
plt.subplot(2,1,1)
plt.plot(df.국어, 'r-*', lw=1)
plt.subplot(2,1,2)
plt.plot(df.영어, 'b--o', lw=1)
plt.tight_layout() #그래프 간격을 겹치지 않게 조절함
```

```python
#Method 2 객체지향 API를 이용
import matplotlib.pyplot as plt
fig = plt.figure()

ax = fig.add_subplot(2,1,1)
ax.plot(df.국어, 'r-*', lw=1)
ax = fig.add_subplot(2,1,2)
ax.plot(df.영어, 'b--o', lw=1)
plt.tight_layout() #그래프 간격을 겹치지 않게 조절함
```

```python
#Method 3 subplots()를 이용
import matplotlib.pyplot as plt
fig,ax = plt.subplots(2,1) #2행 1열 ax를 한꺼번에 생성

ax[0].plot(df.국어, 'r-*', lw=1)
ax[1].plot(df.영어, 'b--o', lw=1)
plt.tight_layout() #그래프 간격을 겹치지 않게 조절함
```

subplots()를 이용

```python
fig,ax=plt.subplots(2,2) #2행 2열 ax를 한꺼번에 생성
ax[0,0].plot(df.국어,'r')
ax[0,1].plot(df.영어,'g')
ax[1,0].plot(df.수학,'b')
ax[1,1].plot(df.과학,'k')
```

x축 라벨을 45도로 기울이고, 그래프 간격을 조정하는 옵션 설정

```python
fig,ax=plt.subplots(2,2) #2행 2열 ax를 한꺼번에 생성
ax[0,0].plot(df.국어,'r-*')
ax[0,0].tick_params(axis='x', rotation=45)
ax[0,1].plot(df.영어,'g-^')
ax[0,1].tick_params(axis='x', rotation=45)
ax[1,0].plot(df.수학,'b-o')
ax[1,0].tick_params(axis='x', rotation=45)
ax[1,1].plot(df.과학,'s-', color='black')
ax[1,1].tick_params(axis='x', rotation=45)
plt.tight_layout() #그래프 간격을 겹치지 않게 조절함
```
